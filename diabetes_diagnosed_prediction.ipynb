# **Split Train Data**

```python
X_train = df.drop("diagnosed_diabetes", axis=1)
y_train = df["diagnosed_diabetes"]
```

# **X Test**

```python
X_test = df_test.copy()
```

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
from sklearn.preprocessing import OneHotEncoder, OrdinalEncoder
```

```python
rank_cat=['education_level','income_level']
un_rank_cat=['gender','ethnicity','smoking_status','employment_status']
```

# **xgboost**

```python
from xgboost import XGBClassifier

xgb_clf = XGBClassifier(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=10,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="binary:logistic",
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)
```

# **FunctionTransformer**

```python
from sklearn.preprocessing import FunctionTransformer
```

```python
log_transform=FunctionTransformer(np.log1p,validate=False)
```

```python
transform_cols=['age', 'physical_activity_minutes_per_week', 'diet_score','bmi', 'systolic_bp', 'diastolic_bp', 'heart_rate', 'cholesterol_total', 'hdl_cholesterol', 'ldl_cholesterol', 'triglycerides']
```

# **Pipeline**

```python
pipe=Pipeline([
   ('CL',ColumnTransformer([
       ('f_t',log_transform,transform_cols),
       ('cat',OneHotEncoder(drop='first'),un_rank_cat),
       ('rank',OrdinalEncoder(),rank_cat)
   ])),
  ('model',xgb_clf )
])
```

```python
from sklearn.model_selection import StratifiedGroupKFold
```

```python
sgkf = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
from sklearn.model_selection import cross_validate

scores = cross_validate(
    pipe, X_train, y_train,
    cv=sgkf,
    groups=X_train['id'],
    scoring=['accuracy', 'precision', 'recall', 'f1'],
    return_train_score=True
)
```

```python
scores
```

```python
pipe.fit(X_train, y_train)
```

```python
y_pred=pipe.predict(X_test)
```

# **Kaggle submission**

```python
submission = pd.DataFrame({
    'id': df_test['id'],
    'target': y_pred
})

submission.to_csv('diabetes_diagnosed_submission.csv', index=False)
```

# **lightgbm**

```python
from lightgbm import LGBMClassifier

lgbm_clf = LGBMClassifier(
    n_estimators=500,
    learning_rate=0.05,
    num_leaves=31,
    max_depth=10,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="binary",
    random_state=42,
    n_jobs=-1
)
```

# **Second Pipeline**

```python
pipe_2=Pipeline([
   ('CL',ColumnTransformer([
       ('f_t',log_transform,transform_cols),
       ('cat',OneHotEncoder(drop='first'),un_rank_cat),
       ('rank',OrdinalEncoder(),rank_cat)
   ])),
  ('model',lgbm_clf )
])
```

```python
sgkf_2 = StratifiedGroupKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
scores_2 = cross_validate(
    pipe_2, X_train, y_train,
    cv=sgkf_2,
    groups=X_train['id'],
    scoring=['accuracy', 'precision', 'recall', 'f1'],
    return_train_score=True
)
```

```python
scores_2
```

```python
pipe_2.fit(X_train, y_train)
```

```python
y_pred_2=pipe_2.predict(X_test)
```

# **Kaggle submission**

```python
submission_2 = pd.DataFrame({
    'id': df_test['id'],
    'diagnosed_diabetes': y_pred_2
})

submission_2.to_csv('diabetes_diagnosed_submission_light_boost.csv', index=False)
```
